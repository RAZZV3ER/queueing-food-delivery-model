```python
import numpy as np
import matplotlib.pyplot as plt

# Service rate (fixed)
mu = 10  # orders per hour

# Arrival rates (varying)
lambda_values = np.linspace(1, 9, 50)

# Waiting time formula
W = 1 / (mu - lambda_values)

# Plot
plt.plot(lambda_values, W)
plt.xlabel("Arrival Rate (orders/hour)")
plt.ylabel("Waiting Time (hours)")
plt.title("Waiting Time vs Arrival Rate (M/M/1 Queue)")
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

mu = 10
lambda_values = np.linspace(1, 9, 50)

L = lambda_values / (mu - lambda_values)

plt.plot(lambda_values, L)
plt.xlabel("Arrival Rate")
plt.ylabel("Average Number of Orders (L)")
plt.title("Queue Length vs Arrival Rate")
plt.grid()
plt.show()
```

```python
import random
import matplotlib.pyplot as plt

arrival_rate = 9
service_rate = 8

queue = 0
queue_history = []

for i in range(100):
    # arrivals based on rate
    arrivals = random.randint(1, 5)   # increase randomness
    queue += arrivals

    # service (realistic)
    served = min(queue, random.randint(1, 3))
    queue -= served

    queue_history.append(queue)

plt.plot(queue_history)
plt.xlabel("Time Steps")
plt.ylabel("Queue Length")
plt.title("Queue Build-Up Over Time (Overloaded System)")
plt.grid()
plt.show()
```
